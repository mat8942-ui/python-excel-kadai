```python
import openpyxl
import time
# openpyxl.stylesモジュールのFontクラスをインポート
from openpyxl.styles import Font

# Workbookクラスのインスタンス化
workbook = openpyxl.Workbook()
# アクティブなシートを変数「ws」に格納する
ws = workbook.active

# セルに値を書き込む
ws["B2"] = "請求書"
ws["B4"] = "株式会社ABC"
ws["B5"] = "〒101-0022 東京都千代田区神田練塀町300"
ws["B6"] = "TEL:03-1234-5678 FAX:03-1234-5678"
ws["B7"] = "担当者名:鈴木一郎 様"
ws["F4"] = "No."
ws["G4"] = "0001"
ws["F5"] = "日付"
ws["G5"] = time.strftime("%Y/%m/%d",time.localtime())
ws["B14"] = "小計"
ws["B15"] = "消費税"
ws["B16"] = "合計"


category = ['商品名','数量', '単価','金額']

row = 0

for i , cat in enumerate(category):
  ws.cell(row = 10, column = 2+i, value = cat)


Product_group = ['商品A','商品B']
count = ['2','1']
price = ['10000','15000']


for i, name in enumerate(Product_group):
  row = 11 + i
  ws.cell(row = row, column = 2 , value = name)
  ws.cell(row = row, column = 3 , value = count[i])
  ws.cell(row = row, column = 4 , value = price[i])
  ws.cell(row = row, column = 5 , value = int(count[i])*int(price[i]))

total = 0

for row in ws.iter_rows(min_row = 11,max_row = 12,min_col = 5, max_col = 5,values_only = True):
  total += row[0]

ws.cell(row = 14, column = 5, value = total)
ws.cell(row = 15, column = 5, value = int(total)*0.1 )
ws.cell(row = 16, column = 5, value = int(total)*1.1 )

workbook.save("請求書.xlsx")
















```
