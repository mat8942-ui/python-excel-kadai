```python
import pandas as pd

#データフレームの作成

df = pd.DataFrame({'日付':['2023-05-17','2023-05-18','2023-05-19','2023-05-20','2023-05-21'],
'社員名':['山田','佐藤','鈴木','田中','高橋'],
'売上':['100','200','150','300','250'],
'部門':['メーカー','代理店','メーカー','商社','代理店'],
                   })

#売上平均値列の作成
df['売上'] = pd.to_numeric(df['売上'])
df['売上平均値']  = df['売上'].mean()

average = df['売上'].mean()

#業績ランク列の作成
def Performance(level):

  result = ''
  if level >= average +50:
    result = 'A'
  elif level >= average :
    result = 'B'
  elif level <= average :
    result = 'C'

  return result

df['業績ランク'] = df['売上'].apply(Performance)


  #エクセルファイルの作成
writer = pd.ExcelWriter('業績.xlsx')

  #データフレームオブジェクトをエクセルファイルに書きこむ
df.to_excel(writer,sheet_name='sheet1',index=False)

writer.close()





```
