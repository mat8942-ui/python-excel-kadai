```python
import openpyxl
import pandas as pd
from openpyxl.chart import (

    Reference,
    Series,
)
```

```python
#エクセルファイルを読み込む
df1 = pd.read_excel('2022_年間売上表.xlsx')
df2 = pd.read_excel('2023_年間売上表.xlsx')


#処理対象のファイル名のリストを格納
file_list = [df1,df2]

df = pd.DataFrame()


#ファイルをひとつづつ処理する
for file in file_list:
  df = pd.concat([df,file],ignore_index=True)

result = df.groupby(['商品', '売上年'])['金額（千円）'].sum().reset_index()
result.columns = ['商品', '売上年', '金額（千円）']
print(result)

result.to_excel('売上集計表.xlsx', index=False)
from openpyxl import load_workbook

wb = load_workbook('売上集計表.xlsx')
ws = wb.active
from openpyxl.styles import PatternFill

for cell in ws[1]:
    cell.fill = PatternFill(
        fill_type='solid',
        fgColor='F2F2F2'
    )

    wb.save('売上集計表.xlsx')




```
